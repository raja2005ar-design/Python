```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Create a sample DataFrame 'df' with 'study_hours' and 'marks' columns
data = {'study_hours': np.random.rand(10) * 10,  # Sample study hours between 0 and 10
        'marks': np.random.rand(10) * 100}   # Sample marks between 0 and 100
df = pd.DataFrame(data)

plt.figure(figsize=(8,5))
sns.lineplot(
    x="study_hours",
    y="marks",
    data=df,
    marker="o")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
plt.figure(figsize=(8,5))
sns.scatterplot(
    x="study_hours",
    data=df,
    y="marks",
    s=100)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
plt.figure(figsize=(8,5))
sns.barplot(
    x="study_hours",
    y="marks",
    data=df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Create a sample DataFrame 'df' with 'study_hours', 'marks', and 'department' columns
data = {'study_hours': np.random.rand(10) * 10,
        'marks': np.random.rand(10) * 100,
        'department': np.random.choice(['Science', 'Arts', 'Commerce'], 10)}
df = pd.DataFrame(data)

plt.figure(figsize=(8,5))
sns.countplot(
    x="department",
    data=df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
plt.figure(figsize=(8,5))
sns.histplot(
    df["marks"],
    bins=5,
    )
plt.title("Distribution of student marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
plt.figure(figsize=(8,5))
sns.boxplot(
    x="department",
    y="marks",
    data=df)
plt.title("box plot of student marks")
plt.xlabel("marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
plt.figure(figsize=(8,5))
sns.violinplot(
    x="department",
    y="marks",
    data=df)
plt.title("marks distribution by department")
plt.xlabel("department")
plt.ylabel("marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Add a sample 'Age' column to df for the pairplot
df['Age'] = np.random.randint(18, 25, size=len(df)) # Sample ages between 18 and 24

plt.figure(figsize=(8,5))
sns.pairplot(
    df[["Age", "study_hours", "marks"]]
)
plt.show()
```
